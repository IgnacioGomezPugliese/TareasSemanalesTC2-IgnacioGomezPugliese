```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import lfilter, freqz

fs = 1000.0
Ts = 1.0/fs
f0 = 5.0
t = np.arange(0, 1, Ts)
x = np.sin(2*np.pi*f0*t)

# ---------- 1) implementación con lfilter ----------
# H(z) = 0.5*(1 - z^-2)
b = [0.5, 0.0, -0.5]
a = [1.0]
y_lf = lfilter(b, a, x)        # salida según coeficientes
y_lf_dt = y_lf / Ts            # pasar a unidades /s (derivada aproximada)

# ---------- 2) implementación usando diff ----------
# d[n] = x[n] - x[n-1]; usamos prepend=0 para que d tenga mismo largo que x
d = np.diff(x, prepend=0.0)    # d[0] = x[0] - 0
# x[n] - x[n-2] = d[n] + d[n-1]  => construyo y usando esa identidad
y_diff = 0.5 * (d + np.concatenate(([0.0], d[:-1])))
# manejo de bordes: mantener los dos primeros valores en 0 (como en el ejemplo anterior)
y_diff[:2] = 0.0
y_diff_dt = y_diff / Ts

# ---------- 3) derivada ideal ----------
y_ideal = 2*np.pi*f0*np.cos(2*np.pi*f0*t)

# ---------- 4) plots temporales ----------
plt.figure(figsize=(10,5))
plt.plot(t, y_ideal, label='Derivada ideal', linewidth=2)
plt.plot(t, y_lf_dt, '--', label='Diferenciador (lfilter)')
plt.plot(t, y_diff_dt, ':', label='Diferenciador (diff)')
plt.xlim(0, 0.3)
plt.xlabel('Tiempo [s]')
plt.ylabel('Amplitud')
plt.title('Comparación: derivada ideal vs diferenciador discreto')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()



```
